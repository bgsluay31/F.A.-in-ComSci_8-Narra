```python
import math # imports the math lirary

#Input
x1= float(input("Enter x1: "))
y1= float(input("Enter y1: "))


x2= float(input("Enter x2: "))
y2= float(input("Enter y2: "))

#Calculate the distance
Distance = math.sqrt(math.pow(x2-x1,2) + math.pow(y2-y1,2))

#Output
print("The distance between the two points is:", round(Distance,2))



# 1. Using a library helped me simplify my program because instead of making custom math algorithms or making calculations from scratch, we have functions that can significantly make people lives easier in coding. It takes more time to code manuually than just using the library functions.
# 2. The functions that are easier to use are the math.pow and math.sqrt because it makes square roots and powers easier to apply. Imagine you have to apply it to every single one, it would be much more time consuming and difficult right? That's why this library is very convenient for coders who actually lacks time in coding but still wants to make it understandable and precise.
# 3. The program would be more difficult without sqrt() and pow() because you need to manually write all the algorithms. Which will result to a much more increase in code length and complexity.
```
